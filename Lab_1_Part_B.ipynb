### Task 4 — The Five Vs of Big Data

**1. Volume**

The CSV file has a size of 15,115 bytes.

This dataset is not big data by itself because it contains only
305 records and occupies a very small amount of storage space.

It can easily be handled using ordinary tools such as Python,
pandas, or spreadsheet software.

**2. Velocity**

If a real bank receives 50 new loan applications every day:

Number of applications per year = 50 × 365

= 18,250 applications per year.

Therefore, the bank would accumulate approximately 18,250 new
applications annually.

At this scale, a spreadsheet can still manage the data.
However, as the dataset grows and requires more complex processing,
multiple users, and frequent updates, a database or dedicated
data-management system would become more appropriate.

**3. Variety**

The dataset contains structured information organized into rows
and columns.

However, a real bank loan application could also include
unstructured or semi-structured information such as:

- Scanned identification documents.
- Bank statement PDFs.
- Images of collateral.
- Written loan application documents.

These additional data sources are not included in the current dataset.

**4. Veracity**

The dataset contains data-quality problems that may affect
the reliability of analysis.

There are 13 missing values in `monthly_income_etb`,
6 missing values in `has_collateral`, and 5 exact duplicate rows.

These problems demonstrate the importance of checking and cleaning
data before using it for decision-making.

**5. Value**

A bank could use this dataset to evaluate whether requiring
collateral helps reduce loan defaults.

The data shows that applicants without collateral have a higher
default rate than applicants with collateral.

This information could support decisions about collateral
requirements and loan approval policies.

However, the observed relationship alone does not prove that
collateral directly causes lower default rates.

```python
p_default = (df["defaulted"] == "Yes").mean()

print("P(Default):", p_default)

rate_by_collateral = df.groupby("has_collateral")["defaulted"].apply(
    lambda x: (x == "Yes").mean()
)

print(rate_by_collateral)
```

### Task 5 — Conditional Probability, Revisited

**1. Report the probabilities**

The overall probability of default is:

P(Default) = 0.1934

Therefore:

P(Default) = 19.34%

The conditional probability of default for applicants with collateral is:

P(Default | Has Collateral) = 0.1208

Therefore:

P(Default | Has Collateral) = 12.08%

The conditional probability of default for applicants without collateral is:

P(Default | No Collateral) = 0.2600

Therefore:

P(Default | No Collateral) = 26.00%

**2. Does the same pattern from the Unit 1 lecture hold here?**

Yes. The same pattern holds in this dataset.

Applicants without collateral have a default rate of 26.00%,
while applicants with collateral have a default rate of 12.08%.

The difference is:

26.00% - 12.08% = 13.92 percentage points.
